## 🔹 Introduction to Backpropagation

In the previous session, you learned how information flows through a neural network in the **forward direction (feedforward)**. In this session, you will study how a neural network is **trained** using a powerful algorithm called **backpropagation**.

Backpropagation helps the network **learn from its mistakes** by adjusting weights and biases to minimize the error between predicted and actual outputs. By the end of this session, you will be able to **build and train a neural network from scratch using NumPy**, which is also the core objective of this module’s assignment.
