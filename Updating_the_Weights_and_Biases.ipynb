# 🔹 Updating the Weights and Biases - I

In this segment, we introduce **backpropagation** using a simple 3-layer neural network:

- **Network Topology**:
  - Input layer: \(x = [$x_1$, $x_2$]\)
  - 2 Hidden layers
  - 1 Output layer

### **Question**

How are the network topology, loss function, and backpropagation steps organized for a 3-layer neural network designed to classify three categories using one-hot encoded ground truth?

---

### **Answer**

#### **1. Network Topology and Parameters**
The network consists of an input layer, two hidden layers, and one output layer.

* **Input Layer**: Takes a vector $\vec{x} = [x_1, x_2]$.
* **Hidden Layer 1**: Utilizes weight matrix $W^{[1]}$ and bias $b^{[1]}$.
    $$W^{[1]} = \begin{bmatrix} w^{[1]}_{11} & w^{[1]}_{12} \\ w^{[1]}_{21} & w^{[1]}_{22} \end{bmatrix}, \quad b^{[1]} = \begin{bmatrix} b^{[1]}_1 \\ b^{[1]}_2 \end{bmatrix}$$
* **Hidden Layer 2**: Utilizes weight matrix $W^{[2]}$ and bias $b^{[2]}$.
    $$W^{[2]} = \begin{bmatrix} w^{[2]}_{11} & w^{[2]}_{12} \\ w^{[2]}_{21} & w^{[2]}_{22} \end{bmatrix}, \quad b^{[2]} = \begin{bmatrix} b^{[2]}_1 \\ b^{[2]}_2 \end{bmatrix}$$
* **Output Layer**: Utilizes weight matrix $W_o$ and bias vector $b_o$.
    $$W_o = \begin{bmatrix} w_{o11} & w_{o12} \\ w_{o21} & w_{o22} \\ w_{o31} & w_{o32} \end{bmatrix}, \quad b_o = \begin{bmatrix} b_{o1} \\ b_{o2} \\ b_{o3} \end{bmatrix}$$



#### **2. Loss Function**
The network produces a prediction vector $\vec{p} = [p_1, p_2, p_3]^T$ which is compared to a one-hot ground truth vector $y$. The objective is to minimize the **average loss** $G(w, b)$ over $m$ data points:
$$G(w, b) = \frac{1}{m} \sum_{i=1}^{m} L(F(x_i), y_i)$$

#### **3. Backpropagation Step Order**
One iteration of the training process follows these chronological steps:
1.  **Feedforward**: Pass the $i$-th data point through the network to generate a prediction.
2.  **Compute Individual Loss**: Calculate the loss $L$ for that specific data point.
3.  **Aggregate**: Compute the average of all $m$ losses ($G(w, b)$).
4.  **Compute Gradients**: Calculate the gradient of the loss with respect to all weights and biases using the backpropagation algorithm.
5.  **Update Parameters**: Adjust the weights and biases using the calculated gradients.



 **Note:** Even though the last layer's bias may be omitted in some problems, it is defined here for clarity. The output layer produces vector:
\[
p =
\begin{bmatrix}
p_1 \\
p_2 \\
p_3
\end{bmatrix}
\]
to be compared with the **ground truth one-hot vector** \(y\), e.g.,
\[
y \in
\left\{
\begin{bmatrix}1\\0\\0\end{bmatrix},
\begin{bmatrix}0\\1\\0\end{bmatrix},
\begin{bmatrix}0\\0\\1\end{bmatrix}
\right\}.
\]

---

## 🔹 Backpropagation Step Order

In **one iteration of backpropagation**, the computations for a training set of \(m\) points are as follows:

1. **Feedforward** the \(i\)-th data point through the network.
2. **Compute the loss** of the \(i\)-th data point.
3. **Aggregate** the losses over all \(m\) points (average loss \(G(w, b)\)).
4. **Compute gradients** of loss w.r.t. weights and biases using backpropagation.
5. **Update the weights and biases** using the gradients.

---

> This segment sets the foundation for understanding how the gradient of the loss propagates **from the output layer back to the input layers**, which will be detailed in the next segments.

### **Question 1**

Backpropagation: Say you have $m$ data points in the training set. Arrange the following computation steps as they appear in one iteration of the backpropagation algorithm:

1. Feedforward the $i^{th}$ data point
2. Compute the gradient of loss with respect to weights and biases
3. Update the weights and biases
4. Compute the loss of the $i^{th}$ data point
5. Aggregate (compute the average of) $m$ losses

* 1, 4, 2, 5, 3
* **1, 4, 5, 2, 3**
* 1, 5, 4, 2, 3
* 1, 4, 5, 3, 2

---

### **Answer**

The correct sequence is **1, 4, 5, 2, 3**.

The data points are first fed forward, the loss of each data point is computed, then aggregated (to compute (average loss), then the gradient of loss is computed, and finally weights and biases are updated once.


#### **Step-by-Step Explanation**



1.  **Feedforward the $i^{th}$ data point (Step 1)**: The process begins by passing a data point through the network to generate a prediction.
2.  **Compute the loss of the $i^{th}$ data point (Step 4)**: Once the prediction is made, it is compared to the ground truth to calculate the error (loss) for that specific point.
3.  **Aggregate $m$ losses (Step 5)**: After processing the data points, the individual losses are averaged to find the total global loss $G(w, b)$ for the entire set.
4.  **Compute the gradient of loss (Step 2)**: With the total loss calculated, the algorithm performs the actual "backpropagation" to determine how much each weight and bias contributed to the error.
5.  **Update the weights and biases (Step 3)**: Finally, the parameters are adjusted (iterated) using the gradients to reduce the loss in the next cycle.



---

### **Summary Table**

| Order | Action | Description |
| :--- | :--- | :--- |
| **1st** | **1** | Feedforward |
| **2nd** | **4** | Individual Loss Calculation |
| **3rd** | **5** | Loss Aggregation (Average) |
| **4th** | **2** | Gradient Computation |
| **5th** | **3** | Parameter Update |

---

```python
import numpy as np

# Simple dataset: OR gate
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])
y = np.array([[0],[1],[1],[1]])

# Sigmoid activation
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Parameters
input_dim = 2
output_dim = 1
lr = 0.1
epochs = 1000

# Initialize weights and bias
W = np.random.rand(input_dim, output_dim)
b = np.random.rand(1, output_dim)

# Gradient Descent Variants
def train_gd(X, y, W, b, lr, epochs, batch_type="batch", batch_size=2):
    m = X.shape[0]

    for epoch in range(epochs):
        if batch_type == "stochastic":
            indices = np.random.permutation(m)
            X_shuffled = X[indices]
            y_shuffled = y[indices]
            for i in range(m):
                xi = X_shuffled[i:i+1]
                yi = y_shuffled[i:i+1]

                # 1. Feedforward
                z = np.dot(xi, W) + b
                a = sigmoid(z)

                # 4. Compute loss (optional, scalar)
                loss = a - yi

                # 2. Compute gradient
                dz = loss * sigmoid_derivative(a)
                dW = np.dot(xi.T, dz)
                db = dz

                # 3. Update weights
                W -= lr * dW
                b -= lr * db

        else:  # batch or mini-batch
            if batch_type == "batch":
                batches = [(X, y)]
            else:  # mini-batch
                batches = [(X[i:i+batch_size], y[i:i+batch_size]) for i in range(0, m, batch_size)]

            for Xb, yb in batches:
                # 1. Feedforward
                z = np.dot(Xb, W) + b
                a = sigmoid(z)

                # 4. Compute loss
                loss = a - yb

                # 5. Aggregate losses (mean)
                mean_loss = np.mean(loss, axis=0)

                # 2. Compute gradient
                dz = loss * sigmoid_derivative(a)
                dW = np.dot(Xb.T, dz) / Xb.shape[0]
                db = np.mean(dz, axis=0, keepdims=True)

                # 3. Update weights
                W -= lr * dW
                b -= lr * db

    return W, b

# Train with different methods
W_batch, b_batch = train_gd(X, y, W.copy(), b.copy(), lr, epochs, batch_type="batch")
W_sgd, b_sgd = train_gd(X, y, W.copy(), b.copy(), lr, epochs, batch_type="stochastic")
W_mini, b_mini = train_gd(X, y, W.copy(), b.copy(), lr, epochs, batch_type="mini-batch", batch_size=2)

# Predictions
def predict(X, W, b):
    z = np.dot(X, W) + b
    a = sigmoid(z)
    return (a >= 0.5).astype(int)

print("Batch GD Predictions:\n", predict(X, W_batch, b_batch))
print("Stochastic GD Predictions:\n", predict(X, W_sgd, b_sgd))
print("Mini-Batch GD Predictions:\n", predict(X, W_mini, b_mini))
```

# 🔹 Gradient Descent Variants in Backpropagation

Gradient descent is the optimization technique used to update weights and biases in a neural network. The main difference between **Batch**, **Stochastic**, and **Mini-batch** gradient descent lies in how many data points are used to compute the gradients in each iteration.

---

## 1️⃣ Batch Gradient Descent

- **Definition**: Uses **all $m$ data points** in the training set to compute the gradients.
- **Relation to Steps**:
  - Feedforward all points → Compute individual losses → Aggregate → Compute gradient → Update weights.
  - Steps: 1, 4, 5, 2, 3
- **Pros**: Provides a very accurate estimate of the gradient.
- **Cons**: Can be **slow** for large datasets because all data points are used in each iteration.

---

## 2️⃣ Stochastic Gradient Descent (SGD)

- **Definition**: Uses **one data point** at a time to compute the gradient.
- **Relation to Steps**:
  - Feedforward 1 point → Compute loss → Compute gradient → Update weights immediately.
  - Steps become: 1, 4, 2, 3 (skip aggregation)
- **Pros**: Very fast and can escape local minima due to noisy updates.
- **Cons**: Gradient updates are noisy, may fluctuate and take longer to converge smoothly.

---

## 3️⃣ Mini-Batch Gradient Descent

- **Definition**: Uses a **small batch of $b$ data points** (1 < b < m) to compute gradients.
- **Relation to Steps**:
  - Feedforward batch → Compute individual losses → Aggregate batch loss → Compute gradient → Update weights.
  - Steps: 1, 4, 5, 2, 3 (like batch gradient descent, but only on a subset)
- **Pros**: Combines advantages of both Batch and SGD:
  - Faster than full batch
  - Smoother updates than SGD
- **Common Practice**: Batch sizes like 32, 64, 128 are widely used in deep learning.

---

### 🔹 Summary Table

| Variant | Data Used per Update | Steps Used | Pros | Cons |
| :--- | :--- | :--- | :--- | :--- |
| Batch GD | All $m$ points | 1, 4, 5, 2, 3 | Accurate gradient | Slow on large datasets |
| Stochastic GD | 1 point | 1, 4, 2, 3 | Fast, can escape minima | Noisy, fluctuating loss |
| Mini-Batch GD | Small batch | 1, 4, 5, 2, 3 | Fast + stable updates | Slightly less accurate than full batch |

---

> ✅ **Key Takeaway**: The main difference is **how many data points are used to calculate the gradient** at each iteration. Batch = all, Stochastic = one, Mini-batch = few.
